### Single Target Variable: Bar Plot Comparison

```python
# Select target variable to analyze
target_variable = 'major acute cardiovascular event'

# Load all three model types
df_kg = pd.read_csv('kg_interventions_coverage_100_models_all.csv')
df_no_graph_gt = pd.read_csv('no_graph_interventions_coverage_100_models_gt_order.csv')
df_no_graph_random = pd.read_csv('no_graph_interventions_coverage_100_models_random1.csv')

# Filter by target variable
df_kg_target = df_kg[df_kg['Target Variable'] == target_variable]
df_no_graph_gt_target = df_no_graph_gt[df_no_graph_gt['Target Variable'] == target_variable]
df_no_graph_random_target = df_no_graph_random[df_no_graph_random['Target Variable'] == target_variable]

print(f"Plotting interventions for target variable: {target_variable}")
print(f"KG models: {len(df_kg_target)} interventions")
print(f"No-Graph GT Order: {len(df_no_graph_gt_target)} interventions")
print(f"No-Graph Random: {len(df_no_graph_random_target)} interventions")
```

```python
# Plot comparison: KG vs No-Graph GT Order
plot_single_target_coverage_bars(
    df_kg=df_kg_target,
    df_no_graph=df_no_graph_gt_target,
    target_var=target_variable,
    output_filename=f'coverage_bars_{target_variable.replace(" ", "_")}_kg_vs_gt_order.png'
)
```
